**3. Quick look and download.**

````python
import pandas as pd
h = pd.read_csv("live_hourly.csv", parse_dates=["timestamp"])
print(h.timestamp.min(), "to", h.timestamp.max(), "|", int(h.calls.sum()), "calls,",
      int(h.calls.isna().sum()), "hours without listening")
h.set_index("timestamp").calls.plot(figsize=(10, 3), title="Right whale calls per hour (WHOI buoy)")
from google.colab import files
files.download("live_hourly.csv")
````
